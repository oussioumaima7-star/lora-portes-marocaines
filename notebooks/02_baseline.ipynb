```
from src.generate import load_pipeline, generate
pipe = load_pipeline()
prompts = ['a photo of a traditional moroccan door',
           'a moroccan door with blue zellige tiles around it',
           'a photo of a mrcdoor']  # mot inconnu -> image quelconque
import matplotlib.pyplot as plt
fig, ax = plt.subplots(len(prompts), 4, figsize=(14, 3.5 * len(prompts)))
for i, p in enumerate(prompts):
    imgs, dt = generate(pipe, p, seed=0, n=4)
    for j, im in enumerate(imgs): ax[i, j].imshow(im); ax[i, j].axis('off')
    ax[i, 0].set_title(f'{p} ({dt:.1f}s)', loc='left', fontsize=9)
plt.tight_layout(); plt.savefig('results/baseline.jpg', dpi=100)
```
